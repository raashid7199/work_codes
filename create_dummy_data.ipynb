```python
import pandas as pd
import random
from datetime import datetime, timedelta
def create_dummy_data(n_rows=5):
    return pd.DataFrame({

        #enter the col names in this dictionaries as per the data working with
        "customer_id": [f"cust{i}" for i in range(n_rows)],
        "email": [f"user{i}@example.com" for i in range(n_rows)],
        "phone": [f"9{random.randint(100000000, 999999999)}" for _ in range(n_rows)],
        "salary": [round(random.uniform(30000, 120000), 2) for _ in range(n_rows)],
        "created_date": [
            (datetime.today() - timedelta(days=random.randint(1, 1000))).strftime("%Y-%m-%d")
            for _ in range(n_rows)
        ]
    })

dummy_df = create_dummy_data()
dummy_df.to_csv("prep_dummy_data.csv", index=False)
```
